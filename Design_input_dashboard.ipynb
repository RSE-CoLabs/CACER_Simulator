# Dashboard users CACER

Esegui la cella seguente per avviare la dashboard Streamlit che costruisce il file `users CACER.xlsx` a partire dal template originale.

```python
from pathlib import Path
import subprocess
import sys

app_path = Path("dashboard/users_cacer_dashboard.py").resolve()
port = "8501"

cmd = [
    sys.executable,
    "-m",
    "streamlit",
    "run",
    str(app_path),
    "--server.address",
    "127.0.0.1",
    "--server.port",
    port,
]

process = subprocess.Popen(cmd)
print(f"Dashboard avviata su http://127.0.0.1:{port}")
print(f"PID processo Streamlit: {process.pid}")
```

Per fermare il server avviato dalla cella precedente, esegui questa cella.

```python
process.terminate()
print("Dashboard fermata.")
```
