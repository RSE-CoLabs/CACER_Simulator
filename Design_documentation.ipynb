# MkDocs setup tutorial

This notebook uses the helper functions defined in `documentation/mkdocs_tools.py` to manage the repository documentation:

1. `run_local_documentation()` builds the MkDocs site and serves it locally.
2. `free_localhost_port()` releases the local host port if it is already occupied.
3. `deploy_github_pages()` publishes the documentation to GitHub Pages on the `gh-pages` branch of the configured Git remote.


## Imports

Run this cell once. The imported functions assume that the notebook is executed from the repository root and use the `documentation/mkdocs.yml` configuration file.

```python
from documentation.mkdocs_tools import (
    deploy_github_pages,
    free_localhost_port,
    run_local_documentation,
)
```

## Local preview

Run the following cell to build the documentation and open it at `http://127.0.0.1:8000/`.

```python
run_local_documentation()
```

## Release the local host port

Run the following cell if `http://127.0.0.1:8000/` is already occupied by a previous local server.

```python
free_localhost_port(port=8000)
```

## Deploy GitHub Pages

Run the following cell when you want to publish the updated documentation online. The function asks for confirmation before pushing to the `gh-pages` branch.

```python
deploy_github_pages()
```

After the deploy, GitHub Pages will be available at the address configured in the repository Pages settings.
